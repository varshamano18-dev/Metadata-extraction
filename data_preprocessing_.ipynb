```python
import os
import hashlib
from collections import defaultdict

train_folder = '/content/train'

file_hashes = defaultdict(list)

for filename in os.listdir(train_folder):
    filepath = os.path.join(train_folder, filename)

    if os.path.isfile(filepath):
        with open(filepath, 'rb') as f:
            file_hash = hashlib.sha256(f.read()).hexdigest()

        file_hashes[file_hash].append(filename)

print("TOTAL FILES:", len(os.listdir(train_folder)))

print("\nDUPLICATE FILES:")
duplicates_found = False

for files in file_hashes.values():
    if len(files) > 1:
        duplicates_found = True
        print(files)

if not duplicates_found:
    print("No duplicate files found.")

print("\nUNIQUE FILES:", len(file_hashes))
```

```python

import pandas as pd
import os

# 1. Load original training dataset
df = pd.read_csv('/content/train.csv')

print("ORIGINAL DATASET SHAPE:", df.shape)

# 2. Remove duplicate records
duplicate_count = df.duplicated().sum()
df = df.drop_duplicates().copy()

print("\nDuplicate records found:", duplicate_count)
print("Duplicate records after cleaning:", df.duplicated().sum())

# 3. Handle missing values in Renewal Notice (Days)
notice_col = 'Renewal Notice (Days)'

df[notice_col] = pd.to_numeric(df[notice_col], errors='coerce')

missing_before = df[notice_col].isnull().sum()
median_value = df[notice_col].median()

df[notice_col] = df[notice_col].fillna(median_value)

print("\nMissing renewal notice values before:", missing_before)
print("Median used:", median_value)
print("Missing renewal notice values after:", df[notice_col].isnull().sum())

# 4. Correct dates using the original rental agreements
file_col = 'File Name'
start_col = 'Aggrement Start Date'
end_col = 'Aggrement End Date'

date_corrections = {
    '50070534': {
        start_col: '01-04-2010',
        end_col: '30-03-2011'
    },
    '18325926': {
        end_col: '04-11-2009'
    },
    '36199312': {
        end_col: '31-03-2011'
    },
    '47854715': {
        end_col: '28-02-2011'
    }
}

print("\nDATE CORRECTIONS:")

for file_id, corrections in date_corrections.items():
    mask = df[file_col].astype(str).str.contains(
        file_id, regex=False, na=False
    )

    if mask.sum() == 1:
        for column, value in corrections.items():
            df.loc[mask, column] = value
        print(file_id, "corrected successfully")
    elif mask.sum() == 0:
        print("WARNING:", file_id, "not found in CSV")
    else:
        print("WARNING:", file_id, "matched multiple records")

# 5. Standardize date format
for column in [start_col, end_col]:
    parsed_dates = pd.to_datetime(
        df[column].astype(str).str.strip(),
        format='mixed',
        dayfirst=True,
        errors='coerce'
    )

    df[column] = parsed_dates.dt.strftime('%d-%m-%Y')

# 6. Final validation
start_dates = pd.to_datetime(
    df[start_col],
    format='%d-%m-%Y',
    errors='coerce'
)

end_dates = pd.to_datetime(
    df[end_col],
    format='%d-%m-%Y',
    errors='coerce'
)

invalid_ranges = (
    start_dates.notna() &
    end_dates.notna() &
    (start_dates > end_dates)
)

print("\n========== FINAL VALIDATION ==========")

print("Final dataset shape:", df.shape)
print("\nDuplicate records:", df.duplicated().sum())
print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nMissing/invalid start dates:", start_dates.isnull().sum())
print("Missing/invalid end dates:", end_dates.isnull().sum())
print("Incorrect date ranges:", invalid_ranges.sum())

if start_dates.isnull().any():
    print("\nRecords with invalid start dates:")
    display(df.loc[start_dates.isnull()])

if end_dates.isnull().any():
    print("\nRecords with invalid end dates:")
    display(df.loc[end_dates.isnull()])

if invalid_ranges.any():
    print("\nRecords with incorrect date ranges:")
    display(df.loc[invalid_ranges])

# 7. Save final cleaned dataset
output_path = '/content/final_cleaned_train.csv'
df.to_csv(output_path, index=False)

print("\nFinal dataset saved successfully!")
print("File:", output_path)
print("Final dataset shape:", df.shape)

# 8. Display final dataset
display(df)
```

```python
import pandas as pd
from google.colab import files
import os

# 1. Load the original uploaded test dataset
test = pd.read_csv('/content/test (1).csv')

# 2. Load the already-cleaned training dataset
train = pd.read_csv('/content/final_cleaned_train.csv')

# 3. Clean column names
test.columns = test.columns.str.strip()
train.columns = train.columns.str.strip()

# 4. Remove duplicate records
test = test.drop_duplicates().copy()

# 5. Remove extra spaces from text values
for col in test.select_dtypes(include='object').columns:
    test[col] = test[col].str.strip()

# 6. Convert date columns into a consistent format
date_columns = [
    col for col in test.columns
    if 'date' in col.lower()
]

for col in date_columns:
    test[col] = pd.to_datetime(
        test[col], dayfirst=True, errors='coerce'
    ).dt.strftime('%d.%m.%Y')

# 7. Convert numeric columns
for col in ['Aggrement Value', 'Renewal Notice (Days)']:
    if col in test.columns:
        test[col] = pd.to_numeric(test[col], errors='coerce')

# 8. Handle missing values using training data
for col in test.columns:
    if test[col].isnull().any() and col in train.columns:

        if pd.api.types.is_numeric_dtype(test[col]):
            train_values = pd.to_numeric(train[col], errors='coerce')
            fill_value = train_values.median()
        else:
            fill_value = train[col].dropna().mode()
            fill_value = fill_value.iloc[0] if not fill_value.empty else None

        if fill_value is not None:
            test[col] = test[col].fillna(fill_value)

# 9. Verify the cleaned test dataset
print("Final Test Data Shape:", test.shape)
print("\nMissing Values:")
print(test.isnull().sum())
print("\nDuplicate Records:", test.duplicated().sum())
print("\nCleaned Test Data:")
print(test.head())

# 10. Save the final cleaned test dataset
output_file = '/content/final_cleaned_test.csv'
test.to_csv(output_file, index=False)

# 11. Download the final file
files.download(output_file)

print("\nFinal cleaned test dataset saved and downloaded successfully!")
```

```python

import os
import pandas as pd
from docx import Document
from PIL import Image
import pytesseract

# Read the already cleaned training metadata
train = pd.read_csv('/content/final_cleaned_train.csv')

# Store extracted document text
extracted_documents = []

for root, dirs, files in os.walk('/content/train/train'):
    dirs[:] = [d for d in dirs if d != '__MACOSX']

    for filename in files:
        if filename.startswith('._'):
            continue

        filepath = os.path.join(root, filename)
        text_content = ""

        try:
            if filename.lower().endswith('.docx'):
                doc = Document(filepath)
                text_content = "\n".join(
                    paragraph.text for paragraph in doc.paragraphs
                )

                # Also extract text from tables
                for table in doc.tables:
                    for row in table.rows:
                        text_content += "\n" + " ".join(
                            cell.text for cell in row.cells
                        )

            elif filename.lower().endswith(('.png', '.jpg', '.jpeg')):
                text_content = pytesseract.image_to_string(
                    Image.open(filepath)
                )

            extracted_documents.append({
                'File Name': filename,
                'Extracted Text': text_content
            })

            print("Extracted:", filename)

        except Exception as e:
            print("Error:", filename, e)

extracted_df = pd.DataFrame(extracted_documents)

print("\nTotal extracted documents:", len(extracted_df))
display(extracted_df[['File Name', 'Extracted Text']])
```

```python

import pandas as pd
import re

if 'extracted_df' not in globals() or 'train' not in globals():
    print("ERROR: extracted_df or train is missing.")
else:
    documents = extracted_df.copy()
    metadata = train.copy()

    def normalize_column(name):
        name = str(name).strip().lower()
        name = name.replace("aggrement", "agreement")
        return re.sub(r'[^a-z0-9]', '', name)

    documents.columns = [normalize_column(c) for c in documents.columns]
    metadata.columns = [normalize_column(c) for c in metadata.columns]

    required_fields = {
        'agreementvalue': 'Agreement Value',
        'agreementstartdate': 'Agreement Start Date',
        'agreementenddate': 'Agreement End Date',
        'renewalnoticedays': 'Renewal Notice (Days)',
        'partyone': 'Party One',
        'partytwo': 'Party Two'
    }

    missing = []

    for col in ['filename', 'extractedtext']:
        if col not in documents.columns:
            missing.append("Document column: " + col)

    for col in ['filename'] + list(required_fields.keys()):
        if col not in metadata.columns:
            missing.append("Metadata column: " + col)

    if missing:
        print("Required columns not found:")
        print(missing)
    else:
        documents['Match_ID'] = documents['filename'].astype(str).str.extract(r'^(\d+)')[0]
        metadata['Match_ID'] = metadata['filename'].astype(str).str.extract(r'^(\d+)')[0]

        documents = documents.dropna(subset=['Match_ID'])
        metadata = metadata.dropna(subset=['Match_ID'])

        metadata = metadata.rename(columns=required_fields)
        target_columns = list(required_fields.values())

        training_data = documents.merge(
            metadata[['Match_ID'] + target_columns],
            on='Match_ID',
            how='left',
            validate='many_to_one'
        )

        matched = training_data[target_columns].notna().all(axis=1)

        print("Total extracted documents:", len(documents))
        print("Total metadata records:", len(metadata))
        print("Fully matched documents:", matched.sum())
        print("Unmatched documents:", (~matched).sum())

        print("\nMissing values:")
        print(training_data[target_columns].isnull().sum())

        print("\nMatched training data:")
        display(training_data.loc[matched])

        print("\nUnmatched documents:")
        display(training_data.loc[~matched, ['filename', 'Match_ID']])

        print("\nFinal training data shape:", training_data.shape)

        if matched.all() and len(training_data) > 0:
            print("\nSUCCESS: All documents have complete metadata.")
        else:
            print("\nCHECK REQUIRED: Some metadata is still unmatched.")
```
