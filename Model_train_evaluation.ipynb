```python
import glob, os, json, docx, pytesseract
from PIL import Image

paths = {}
for p in glob.glob('/content/files/**/*.*', recursive=True):
    if p.lower().endswith(('.docx', '.png')):
        paths[os.path.splitext(os.path.basename(p))[0]] = p

def find_path(file_name):
    return paths[os.path.splitext(str(file_name))[0]]

def read_docx(path):
    d = docx.Document(path)
    lines = [para.text for para in d.paragraphs]
    for table in d.tables:
        for row in table.rows:
            lines.append(' | '.join(cell.text for cell in row.cells))
    return '\n'.join(lines)

def read_doc(path):
    if path.lower().endswith('.docx'):
        text = read_docx(path)
    else:
        text = pytesseract.image_to_string(Image.open(path))
    return text[:6000]
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import LoraConfig, get_peft_model

model_name = 'Qwen/Qwen2.5-1.5B-Instruct'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, dtype=torch.float32).to('cuda')

model.config.use_cache = False
model.gradient_checkpointing_enable()
model.enable_input_require_grads()

config = LoraConfig(r=16, lora_alpha=32, lora_dropout=0.05,
                    target_modules=['q_proj', 'k_proj', 'v_proj', 'o_proj'],
                    task_type='CAUSAL_LM')
model = get_peft_model(model, config)
model.print_trainable_parameters()
```

```python
instruction = (
    "Extract these 6 fields from the agreement: "
    "Agreement Value, Agreement Start Date, Agreement End Date, "
    "Renewal Notice (Days), Party One, Party Two. "
    "Reply only with a JSON object using exactly these keys."
)

def make_prompt(path):
    messages = [
        {'role': 'system', 'content': 'You extract metadata from agreements.'},
        {'role': 'user', 'content': instruction + '\n\n' + read_doc(path)},
    ]
    return tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

def make_example(row):
    prompt_ids = tokenizer(make_prompt(find_path(row[name_col])))['input_ids']
    answer = json.dumps({c: row[c] for c in field_cols}) + tokenizer.eos_token
    answer_ids = tokenizer(answer)['input_ids']
    x = torch.tensor([prompt_ids + answer_ids]).to('cuda')
    y = torch.tensor([[-100] * len(prompt_ids) + answer_ids]).to('cuda')
    return x, y

x, y = make_example(train.iloc[0])
print(tokenizer.decode(y[0][y[0] != -100]))
```

```python
opt = torch.optim.AdamW([p for p in model.parameters() if p.requires_grad], lr=2e-4)

model.train()
for epoch in range(6):
    total = 0
    for i in range(len(train)):
        x, y = make_example(train.iloc[i])
        with torch.autocast('cuda', dtype=torch.float16):
            loss = model(input_ids=x, labels=y).loss
        loss.backward()
        opt.step()
        opt.zero_grad()
        total += loss.item()
    print('epoch', epoch + 1, 'loss', round(total / len(train), 4))
```

```python
def predict(path):
    x = tokenizer(make_prompt(path), return_tensors='pt').to('cuda')
    with torch.no_grad():
        out = model.generate(**x, max_new_tokens=150, do_sample=False, use_cache=True)
    reply = tokenizer.decode(out[0][x['input_ids'].shape[1]:], skip_special_tokens=True)
    try:
        data = json.loads(reply[reply.find('{'):reply.rfind('}') + 1])
        return {c: str(data.get(c, '')) for c in field_cols}
    except:
        return {c: '' for c in field_cols}

def run_all(df):
    rows = []
    for file_name in df[name_col]:
        result = predict(find_path(file_name))
        result[name_col] = file_name
        rows.append(result)
    return pd.DataFrame(rows)[[name_col] + field_cols]

def show_recall(actual, predicted):
    scores = []
    for c in field_cols:
        correct = 0
        for a, p in zip(actual[c], predicted[c]):
            if a.strip().lower() == p.strip().lower():
                correct += 1
        score = correct / len(actual)
        scores.append(score)
        print(c, ':', round(score, 2))
    print('Average recall:', round(sum(scores) / len(scores), 2))
```

```python
import os, re, glob

def norm(s):
    return re.sub(r'[^a-z0-9]', '', str(s).lower())

paths = {}
for p in glob.glob('/content/files/**/*.*', recursive=True):
    if p.lower().endswith(('.docx', '.png', '.jpg', '.jpeg', '.pdf')):
        paths[norm(os.path.basename(p))] = p

def find_path(file_name):
    n = norm(file_name)
    for k, p in paths.items():
        if k.startswith(n):
            return p
    return None

def run_all(df):
    rows = []
    for file_name in df[name_col]:
        p = find_path(file_name)
        if p:
            result = predict(p)
        else:
            print('skipped (file not found):', file_name)
            result = {c: '' for c in field_cols}
        result[name_col] = file_name
        rows.append(result)
    return pd.DataFrame(rows)[[name_col] + field_cols]

for n in test[name_col]:
    print(n, '->', find_path(n))

model.eval()

test_pred = run_all(test)
test_pred.to_csv('predictions.csv', index=False)

print('TEST RECALL')
show_recall(test, test_pred)

files.download('predictions.csv')
test_pred
```
